```python
from ultralytics import YOLO
import os

# Define the dataset path
dataset_path = r"C:\Users\hp\Desktop\X-ray FSOD"  # Use a raw string to avoid issues with backslashes

# Load the YOLOv8 model
model = YOLO("yolov10s.pt")  # You can use 'yolov8s.pt' or another model variant

# Train the model
results = model.train(
    data=os.path.join(dataset_path, "data.yaml"),  # Path to the data.yaml file
    epochs=60,                                    # Number of epochs
    imgsz=640,                                    # Image size
    batch=16,                                     # Batch size
    name="FSOD Xv8"                       # Name of the training run
)

# Display training results
print(results)
# model.train(data=dataset_path, epochs=50, imgsz=224, device="cuda")
```
