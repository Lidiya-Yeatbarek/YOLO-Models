```python
from ultralytics import YOLO
import os

# Define the dataset path
dataset_path = r"C:\Users\hp\Desktop\X-ray FSOD"  # Use a raw string to avoid issues with backslashes

# Train the model
try:
    # Load YOLO model
    model = YOLO("yolov9s.pt")  # Update to yolov8 model

    # Train the model
    results = model.train(
        data=os.path.join(dataset_path, "data.yaml"),
        epochs=60,
        imgsz=640,
        batch=16,
        name="FSOD Sv9 60",
      # Disable Automatic Mixed Precision
    )

    # Display training results
    print(results)

except Exception as e:
    print(f"Error during training: {e}")

```
