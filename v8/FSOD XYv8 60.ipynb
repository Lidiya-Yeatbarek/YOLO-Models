```python
from ultralytics import YOLO
import os

# Define the dataset path
dataset_path = r"C:\Users\hp\Desktop\X-ray FSOD"  # Use raw string to avoid issues with backslashes

# Load the YOLOv8 model (Fix: Use raw string or double backslashes)
model = YOLO(r"C:\Users\hp\yolov8x.pt")  # Correct way to specify Windows file path

# Train the model
results = model.train(
    data=os.path.join(dataset_path, "data.yaml"),  # Path to the data.yaml file
    epochs=60,  # Number of epochs
    imgsz=640,  # Image size
    batch=16,  # Batch size
    name="FSOD_Xv8"  # Name of the training run (Avoid spaces in names)
)

# Display training results
print(results)
```
